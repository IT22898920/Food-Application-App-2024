```python
# Import all necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D
from scipy.stats import pearsonr, spearmanr
import warnings
warnings.filterwarnings('ignore')

# Set style for better plots
plt.style.use('default')
sns.set_palette("husl")

print("🚀 SE4050 Deep Learning Lab 1 - Python Libraries Tutorial")
print("=" * 60)
```

```python
# ===============================
# TASK 1
# ===============================
print("\n📋 TASK 1")
print("-" * 30)

# Task 1.1: Create a random 4x4 array from exponential distribution
print("1️⃣ Creating 4x4 exponential distribution array...")
exponential_4x4 = np.random.exponential(scale=1.0, size=(4, 4))
print("4x4 Exponential Array:")
print(exponential_4x4)
print(f"Shape: {exponential_4x4.shape}")
```

```python
# Task 1.2: Create 100000x1 array and visualize distributions
print("\n2️⃣ Creating distribution comparison plot...")

# Generate data
exponential_data = np.random.exponential(scale=1.0, size=100000)
uniform_data = np.random.uniform(low=0, high=4, size=100000)  # Scaled to match range
normal_data = np.random.normal(loc=1, scale=0.8, size=100000)  # Centered around 1

# Create the plot
plt.figure(figsize=(12, 8))

# Plot histograms
plt.hist(exponential_data, density=True, bins=100, histtype="step",
         color="red", label="Exponential", linewidth=2, alpha=0.8)
plt.hist(uniform_data, density=True, bins=100, histtype="step",
         color="blue", label="Uniform", linewidth=2, alpha=0.8)
plt.hist(normal_data, density=True, bins=100, histtype="step",
         color="green", label="Normal", linewidth=2, alpha=0.8)

# Customize plot
plt.axis([0, 6, 0, 1.2])
plt.legend(loc="upper right", fontsize=12)
plt.title("Random Distributions Comparison", fontsize=16, fontweight='bold')
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Task 1.3: 3D Plot of Z = X**2 + Y**2
print("\n3️⃣ Creating 3D surface plot...")

# Create data for 3D plot
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

# Create 3D plot
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Surface plot
surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.9,
                      linewidth=0, antialiased=True)

# Customize 3D plot
ax.set_xlabel('X axis', fontsize=12)
ax.set_ylabel('Y axis', fontsize=12)
ax.set_zlabel('Z = X² + Y²', fontsize=12)
ax.set_title('3D Surface Plot: Z = X² + Y²', fontsize=14, fontweight='bold')

# Add color bar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
plt.show()
```

```python
# Task 1.4: Correlation analysis (using sample Pokemon data)
print("\n4️⃣ Correlation analysis with sample data...")

# Create sample Pokemon-like data for demonstration
np.random.seed(42)
pokemon_data = pd.DataFrame({
    'HP': np.random.normal(70, 20, 150),
    'Attack': np.random.normal(80, 25, 150),
    'Defense': np.random.normal(75, 20, 150),
    'Sp_Atk': np.random.normal(70, 25, 150),
    'Sp_Def': np.random.normal(70, 20, 150),
    'Speed': np.random.normal(65, 25, 150)
})

# Ensure positive values
pokemon_data = pokemon_data.abs()

print("Sample Pokemon Stats Data:")
print(pokemon_data.head())

# Calculate correlations
features = ['HP', 'Attack', 'Defense', 'Sp_Atk', 'Sp_Def', 'Speed']
pearson_corr = pokemon_data[features].corr(method='pearson')
spearman_corr = pokemon_data[features].corr(method='spearman')

# Plot correlation heatmaps
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Pearson correlation heatmap
sns.heatmap(pearson_corr, annot=True, cmap='RdBu_r', center=0,
            square=True, fmt='.3f', ax=ax1, cbar_kws={'label': 'Correlation'})
ax1.set_title('Pearson Correlation Coefficient', fontsize=14, fontweight='bold')

# Spearman correlation heatmap
sns.heatmap(spearman_corr, annot=True, cmap='RdBu_r', center=0,
            square=True, fmt='.3f', ax=ax2, cbar_kws={'label': 'Correlation'})
ax2.set_title('Spearman Rank Correlation', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
print("🚀 Task 2: Au Nanoparticle Dataset Analysis")
print("="*60)

# Load the uploaded dataset
try:
    au_data = pd.read_csv('Au_nanoparticle_dataset.csv')
    print("✅ Successfully loaded Au_nanoparticle_dataset.csv")
    print(f"📊 Dataset shape: {au_data.shape}")
    print(f"📋 All columns: {list(au_data.columns)}")

    # Display basic info
    print(f"\n📝 First 5 rows:")
    print(au_data.head())

    print(f"\n📈 Data types:")
    print(au_data.dtypes)

    print(f"\n🔍 Missing values:")
    print(au_data.isnull().sum())

except Exception as e:
    print(f"❌ Error loading dataset: {e}")
    print("Make sure Au_nanoparticle_dataset.csv is uploaded correctly")
```

```python
# Task 2.1: Filter to required columns
required_cols = ['N_total', 'N_bulk', 'N_surface', 'R_avg']

# Check which columns exist
available_cols = [col for col in required_cols if col in au_data.columns]
print(f"\nRequired columns found: {available_cols}")

if len(available_cols) == 4:
    new_df = au_data[available_cols].copy()
    print(f"✅ Filtered dataframe shape: {new_df.shape}")

    # Task 2.2: First 20 samples
    print("\n2️⃣ First 20 samples:")
    print(new_df.head(20))
# Task 2.3: Statistics
    print("\n3️⃣ Statistics:")
    print(new_df.describe())

    # Detailed stats
    for col in new_df.columns:
        print(f"\n{col}:")
        print(f"  Mean: {new_df[col].mean():.4f}")
        print(f"  Std:  {new_df[col].std():.4f}")
        print(f"  Q1:   {new_df[col].quantile(0.25):.4f}")
        print(f"  Q2:   {new_df[col].quantile(0.50):.4f}")
        print(f"  Q3:   {new_df[col].quantile(0.75):.4f}")

    # Task 2.4: Histograms
    print("\n4️⃣ Creating histograms...")
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    colors = ['skyblue', 'lightcoral', 'lightgreen', 'gold']

    for i, col in enumerate(new_df.columns):
        axes[i].hist(new_df[col], bins=30, alpha=0.7, color=colors[i], edgecolor='black')
        axes[i].set_title(f'{col}', fontweight='bold')
        axes[i].set_xlabel('Value')
        axes[i].set_ylabel('Frequency')
        axes[i].grid(True, alpha=0.3)

        # Add mean line
        mean_val = new_df[col].mean()
        axes[i].axvline(mean_val, color='red', linestyle='--', label=f'Mean: {mean_val:.2f}')
        axes[i].legend()

    plt.suptitle('Distribution of Au Nanoparticle Features', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

    # Task 2.5: Pairplot
    print("\n5️⃣ Creating pairplot...")
    pairplot = sns.pairplot(new_df, diag_kind='hist', plot_kws={'alpha': 0.6})
    pairplot.fig.suptitle('Pairplot of Au Nanoparticle Features', y=1.02, fontsize=16, fontweight='bold')
    plt.show()

    # Task 2.6: Custom PairGrid
    print("\n6️⃣ Creating custom PairGrid...")
    g = sns.PairGrid(new_df, height=3)

    # Upper: bivariate histograms
    g.map_upper(sns.histplot)

    # Diagonal: histogram + KDE
    g.map_diag(sns.histplot, kde=True, stat='density', alpha=0.7)

    # Lower: bivariate KDE
    g.map_lower(sns.kdeplot, fill=True, alpha=0.6)

    g.fig.suptitle('Custom PairGrid: Au Nanoparticle Features\n'
                   'Diagonal: Hist+KDE | Lower: Bivariate KDE | Upper: Bivariate Hist',
                   y=1.02, fontsize=14, fontweight='bold')
    plt.show()

    print("\n✅ Task 2 completed successfully!")

else:
    print(f"❌ Found {len(available_cols)} columns, need 4")
    print(f"Available columns: {list(au_data.columns)}")

print("\n🎉 Analysis Complete!")
```
